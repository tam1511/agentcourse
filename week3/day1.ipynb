```python
async def main():
    agent = Agent(
        name="Trợ lý tư vấn",
        instructions="Bạn là trợ lý tư vấn sản phẩm thân thiện. Trả lời ngắn gọn bằng Tiếng Việt."
    )

    with trace("phiên tư vấn đầu tiên"):
        result = await Runner.run(
            agent,
            "Tôi nên mua điện thoại tầm 5 triệu thì chọn loại nào ?"
        )

    print(result.final_output)

await main()
```

```python
INSTRUCTIONS = """
Bạn là chuyên viên tư vấn sản phẩm của Shopee Việt Nam.

Nhiệm vụ của bạn:
- Giúp khách hàng chọn được sản phẩm phù hợp nhất với nhu cầu và ngân sách
- Hỏi thêm thông tin nếu cần — đừng đoán mò
- Giải thích lý do tại sao sản phẩm/danh mục đó phù hợp, không chỉ liệt kê tên
- Nếu ngân sách quá thấp so với nhu cầu, nói thật và gợi ý giải pháp thực tế

Phong cách: thân thiện, tự nhiên, trực tiếp. Trả lời bằng tiếng Việt.
Độ dài: ngắn gọn, đủ ý — không quá 200 từ mỗi câu trả lời.
"""
```

```python
import asyncio
from agents import Agent, Runner, trace

async def tu_van_san_pham(cau_hoi: str) -> str:

    agent = Agent(
        name="Tư vấn viên Shopee",
        instructions=INSTRUCTIONS,
        model="gpt-4o-mini"
    )
    
    with trace("shopee-tu-van"):
        result = await Runner.run(agent, cau_hoi)

    return result.final_output
```

```python
# test
async def main():
    cac_cau_hoi = [
        "Tôi muốn mua tai nghe để nghe nhạc khi đi làm, ngân sách khoảng 300k",
        "Cho tôi hỏi laptop sinh viên tầm 12-15 triệu thì nên chọn cái nào?",
        "Cần mua quà sinh nhật cho mẹ, bà ấy thích nấu ăn, ngân sách 500k"
    ]

    for cau_hoi in cac_cau_hoi:
        print(f"Khách hỏi: {cau_hoi}")
        tra_loi = await tu_van_san_pham(cau_hoi)
        print(f"Tư vấn viên: {tra_loi}")

await main()
```
