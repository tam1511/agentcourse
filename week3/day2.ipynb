```python
import os
import resend
from dotenv import load_dotenv
from agents import Agent, Runner, trace
```

```python
load_dotenv(override=True)

resend.api_key = os.environ.get("RESEND_API_KEY")
```

```python
from agents import function_tool


@function_tool
def gui_email(noi_dung: str) -> dict:

    params = {
        "from": "onboarding@resend.dev",
        "to": ["thanhtam.udn@gmail.com"],
        "subject": "Giải pháp tăng trưởng cho doanh nghiệp của bạn",
        "text": noi_dung,
    }

    email = resend.Emails.send(params)
    return {
        "status": "success",
        "email_id": email["id"]
    }
```

```python
# Giao tool cho Agent

agent_sales = Agent(
    name="Sales Agent GrowthVN",
    instructions="""Bạn là sales agent của GrowthVN — công ty cung cấp giải pháp 
    marketing automation cho doanh nghiệp Việt Nam.
    
    Khi được yêu cầu gửi email, bạn viết nội dung email ngắn gọn, chuyên nghiệp,
    phù hợp với doanh nghiệp Việt Nam. Sau đó dùng tool để gửi email đó đi.
    
    Phong cách: thân thiện nhưng chuyên nghiệp. Tránh sáo rỗng.
    Độ dài email: không quá 150 từ.""",
    tools=[gui_email],
    model="gpt-4o-mini"
)

async def main():
    with trace("gui-email-test"):
        result = await Runner.run(
            agent_sales,
            "Gửi email giới thiệu dịch vụ GrowthVN đến giám đốc marketing"
        )
    print(result.final_output)

await main()
```

```python
instructions_chuyen_nghiep = """Bạn là sales agent của GrowthVN — công ty cung cấp
giải pháp marketing automation cho SME Việt Nam.
Bạn viết email sales chuyên nghiệp, trang trọng, tập trung vào ROI và hiệu quả kinh doanh."""

instructions_than_thien = """Bạn là sales agent của GrowthVN — công ty cung cấp
giải pháp marketing automation cho SME Việt Nam.
Bạn viết email sales thân thiện, gần gũi, dùng ngôn ngữ tự nhiên của người Việt.
Hài hước nhẹ nhàng khi phù hợp."""

instructions_ngan_gon = """Bạn là sales agent của GrowthVN — công ty cung cấp
giải pháp marketing automation cho SME Việt Nam.
Bạn viết email sales cực kỳ ngắn gọn. Tối đa 5 câu. Đi thẳng vào vấn đề."""

agent_chuyen_nghiep = Agent(
    name="Sales Agent Chuyên Nghiệp",
    instructions=instructions_chuyen_nghiep,
    model="gpt-4o-mini"
)

agent_than_thien = Agent(
    name="Sales Agent Thân Thiện",
    instructions=instructions_than_thien,
    model="gpt-4o-mini"
)

agent_ngan_gon = Agent(
    name="Sales Agent Ngắn Gọn",
    instructions=instructions_ngan_gon,
    model="gpt-4o-mini"
)
```

```python
# chay song song ket qua
import asyncio

async def tao_ba_email_song_song():
    yeu_cau = "Viết email sales giới thiệu dịch vụ GrowthVN đến giám đốc của SME Viêt Nam"

    with trace("ba-email-song-song"):
        ket_qua = await asyncio.gather(
            Runner.run(agent_chuyen_nghiep, yeu_cau),
            Runner.run(agent_than_thien, yeu_cau),
            Runner.run(agent_ngan_gon, yeu_cau),
        )
    
    cac_email = [r.final_output for r in ket_qua]

    for i, email in enumerate(cac_email, 1):
        print(f"\n{'-'*50}")
        print(f"EMAIL #{i}")
        print('-'*50)
        print(email)
    
    return cac_email

await tao_ba_email_song_song()
```

```python
# Agent viet tieu de email
instructions_tieu_de = """Bạn viết tiêu đề (subject line) cho email sales.
Tiêu đề phải hấp dẫn, ngắn gọn (dưới 60 ký tự), khiến người nhận muốn mở email.
Chỉ trả về tiêu đề, không giải thích thêm."""

agent_tieu_de = Agent(
    name="Chuyên gia tiêu đề email",
    instructions=instructions_tieu_de,
    model="gpt-4o-mini"
)

tool_tieu_de = agent_tieu_de.as_tool(
    tool_name="viet_tieu_de",
    tool_description="Viết tiêu đề hấm dẫn cho email sales"
)

# Agent chuyen doi text sang HTML
instructions_html = """Bạn chuyển đổi nội dung email text sang định dạng HTML đẹp.
Giữ nguyên nội dung, thêm định dạng HTML rõ ràng, dễ đọc trên mobile.
Không thêm CSS phức tạp. Chỉ dùng HTML cơ bản."""

agent_html = Agent(
    name="HTML Email Converter",
    instructions=instructions_html,
    model="gpt-4o-mini"
)

tool_html = agent_html.as_tool(
    tool_name="chuyen_html",
    tool_description="Chuyển đổi nội dung text email sang HTML format"
)

# Function tool gui HTML email
@function_tool
def gui_html_email(tieu_de: str, noi_dung_html: str) -> dict:

    params = {
        "from": "onboarding@resend.dev",
        "to": ["thanhtam.udn@gmail.com"],
        "subject": tieu_de,
        "html": noi_dung_html,
    }

    email = resend.Emails.send(params)
    return {
        "status": "success",
        "email_id": email["id"]
    }
```

```python
# Email Sender Agent 
instructions_sender = """Bạn là Email Manager của GrowthVN.
Nhiệm vụ: nhận nội dung email đã được chọn, hoàn thiện và gửi đi.

Quy trình bắt buộc:
1. Dùng tool viet_tieu_de để tạo tiêu đề hấp dẫn
2. Dùng tool chuyen_html để chuyển nội dung sang HTML
3. Dùng tool gui_html_email để gửi email với tiêu đề và HTML vừa tạo

Không bỏ qua bước nào."""

email_sender_agent = Agent(
    name="Email Manager",
    instructions=instructions_sender,
    tools=[tool_tieu_de, tool_html, gui_html_email],
    model="gpt-4o-mini",
    handoff_description="Nhận nội dung email đã chọn, viết tiêu đề, format HTML và gửi đi"
)
```
